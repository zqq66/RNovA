```python
import pandas as pd
from src.fill_PTM import parse_peptide_mods
from workflow import cluster_ptm_pairs_with_counts

df = pd.read_csv('/home/q359zhan/PTM/data/filled_peptides.csv')
ptm_freq = dict()
for row in df.itertuples():
    pep = row.filled_sequence
    ptms = parse_peptide_mods(pep)
    for p in ptms:
        if p['delta_mass'] is not None:
            ptm = p['residue']+'['+str(p['delta_mass']) + ']'
            if ptm in ptm_freq.keys():
                ptm_freq[ptm] += 1
            else:
                ptm_freq[ptm] = 1
clustered_topk_ptm = cluster_ptm_pairs_with_counts(ptm_freq, eps=0.02)
ptm_freq = dict(sorted(clustered_topk_ptm.items(), key=lambda kv: kv[1], reverse=True))

print(ptm_freq)
```
